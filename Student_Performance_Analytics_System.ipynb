```python
import pandas as pd
import numpy as np

# Student Performance Analytics System

# Create student dataset
data = {
    "Student_ID": list(range(101, 121)),
    "Name": [
        "Aarav", "Priya", "Rahul", "Sneha", "Rohan",
        "Neha", "Aditya", "Pooja", "Om", "Anjali",
        "Karan", "Sakshi", "Vivek", "Isha", "Arjun",
        "Meera", "Sahil", "Riya", "Amit", "Kavya"
    ],
    "Department": [
        "IT", "IT", "CS", "CS", "IT",
        "CS", "IT", "CS", "IT", "CS",
        "IT", "CS", "IT", "CS", "IT",
        "CS", "IT", "CS", "IT", "CS"
    ],
    "Python": [
        85, 76, 65, 92, 45, 78, 88, 55, 39, 81,
        72, 95, 61, 48, 90, 67, 42, 83, 58, 97
    ],
    "Mathematics": [
        78, 82, 59, 89, 52, 74, 91, 63, 35, 85,
        68, 93, 57, 44, 86, 72, 48, 79, 62, 96
    ],
    "English": [
        80, 75, 70, 94, 49, 81, 84, 58, 42, 77,
        74, 91, 65, 50, 88, 69, 45, 86, 60, 95
    ],
    "Attendance": [
        92, 85, 78, 96, 65, 88, 94, 72, 58, 90,
        81, 98, 76, 68, 95, 83, 61, 91, 74, 99
    ]
}

df = pd.DataFrame(data)

# Calculate total and average marks
subjects = ["Python", "Mathematics", "English"]
df["Total"] = df[subjects].sum(axis=1)
df["Average"] = df[subjects].mean(axis=1).round(2)

# Function to assign grades
def calculate_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
```
